```python
# Now let's avg all the nutrient data on each seqn_no

# Concatenate the two dataframes
df_nutrients = pd.concat([df_nutrients_day01, df_nutrients_day02])

# List of columns that need to be aggregated
nutrient_columns = [
    'energy_kcal', 'protein_gms', 'carbohydrate_gms', 'total_sugars_gms', 
    'dietary_fiber_gms', 'total_fats_gms', 'saturated_fats_gms', 
    'mono_saturated_fats_gms', 'poly_unsaturated_fats_gms', 'cholestrol_gms', 
    'sodium_mg', 'potassium_mg', 'moisture_gms', 'vitaminC_mg', 
    'vitaminD_mcg', 'vitaK_mcg', 'calcium_mg'
]

# Select only the relevant nutrient columns and group by 'seqn_no'
df_nutrients_filtered = df_nutrients[['seqn_no'] + nutrient_columns]

# Group by 'seqn_no' and calculate the mean for all nutrient columns
df_avg_nutrients = df_nutrients_filtered.groupby('seqn_no').agg('mean').reset_index()

```

```python
# Now building the URL with the values from the .env file
url = f'postgresql://{pg_user}:{pg_pass}@{pg_host}:{pg_port}/{pg_db}'

engine = create_engine(url, echo=False) 
df_avg_nutrients.to_sql(name = 'dietary_nutrient', con=engine, schema='capstone_group_3',if_exists='replace',index=False)
```

```python
# Assuming df_avg_nutrients is the dataframe that contains the average nutrient data

# Create a new dataframe to store the calculated ratios and scores
df_hypertension_risk = pd.DataFrame()

# Include the 'seqn_no' column for reference in the new dataframe
df_hypertension_risk['seqn_no'] = df_avg_nutrients['seqn_no']

# 1. Sodium-to-Potassium Ratio
df_hypertension_risk['sodium_to_potassium_ratio'] = round(df_avg_nutrients['sodium_mg'] / df_avg_nutrients['potassium_mg'],2)

# 2. Fat Quality Score (Saturated Fat, Total Fat vs Mono and Poly Unsaturated Fats)
df_hypertension_risk['fat_quality_score'] = round(df_avg_nutrients['poly_unsaturated_fats_gms'] / (df_avg_nutrients['saturated_fats_gms']),2)

# 3. Cholesterol Score (Directly using cholesterol values)
df_hypertension_risk['cholesterol_score'] = df_avg_nutrients['cholestrol_gms']

# 4. Energy Intake (Energy in kcal)
df_hypertension_risk['energy_intake'] = df_avg_nutrients['energy_kcal']

# 5. Vitamin D Score
df_hypertension_risk['vitaminD_IU'] = df_avg_nutrients['vitaminD_mcg'] * 40

# Get a list of all column names except 'seqn_no'
# These are the columns where you want to check for all null values
columns_to_check_for_nulls = df_hypertension_risk.columns.drop('seqn_no').tolist()

# Drop rows where ALL values in the specified 'subset' of columns are null.
# Using 'inplace=True' modifies the DataFrame directly.
df_hypertension_risk.dropna(subset=columns_to_check_for_nulls, how='all', inplace=True)

# Display the new dataframe with 'seqn_no' and the ratio columns
print(df_hypertension_risk[['seqn_no', 'sodium_to_potassium_ratio', 'fat_quality_score', 'cholesterol_score', 'energy_intake', 'vitaminD_IU']])

```

```python
# Sodium-to-Potassium Ratio:
# - A higher ratio indicates more sodium relative to potassium, which increases blood pressure risk.
# - A lower ratio (less than 1) suggests a healthier balance, which may help lower hypertension risk.

# Fat Quality Score:
# - This score highlights the balance between saturated fats and unsaturated fats.

# - A higher P/S Ratio score(> 1.0) indicates an healthy fat profile beneficial for heart health and lowering blood pressure.
# - A lower P/S Ratio score(< 1.0) reflects a higher intake of saturated fats, which can contribute to hypertension risks.

# Cholesterol Score:
# - Directly uses the cholesterol value (in g) to gauge cardiovascular health.
# - Elevated cholesterol levels are linked to arterial plaque buildup and increased blood pressure.

# Energy Intake:
# - Represents daily calorie intake (in kcal).
# - Excessive energy intake is associated with weight gain, which is a major risk factor for developing hypertension.

# Vitamin D Values:
# - Imbalances in the vitD could negatively impact vascular health.

# These columns provide a comprehensive view of key nutrients linked to hypertension risk, allowing for deeper analysis of dietary factors influencing blood pressure.
```
