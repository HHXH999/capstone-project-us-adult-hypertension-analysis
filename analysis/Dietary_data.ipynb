# Connect data base

```python
# Load config
config = dotenv_values()

# define variables for the login
pg_user = config['POSTGRES_USER']  # align the key label with your .env file !
pg_pass = config['POSTGRES_PASS']
pg_host = config['POSTGRES_HOST']
pg_port = config['POSTGRES_PORT']
pg_db = config['POSTGRES_DB']
pg_schema = config['POSTGRES_SCHEMA']
DB_URL = f"postgresql://{pg_user}:{pg_pass}@{pg_host}:{pg_port}/{pg_db}"

engine = create_engine(DB_URL, echo=False)
```

## Dietary food data for Day 01

```python
# Exporting the Dietary food day 1 data
df_dietary_food_first_day = pd.read_sas('F:\Susanna Course\Capstone_data\DR1IFF_L.xpt', format="xport", encoding="utf-8")
```

```python
df_dietary_food_first_day.head(5)
```

```python
df_dietary_food_first_day.shape
```

```python
# Step 1: Select only the columns with meaningful values
columns_to_keep = ['SEQN', 'DR1CCMNM', 'DR1CCMTX', 'DR1FS', 'DR1IGRMS', 'DR1IKCAL']
df_dietary_food_first_day = df_dietary_food_first_day[columns_to_keep]


# Step 2: Rename the columns
df_dietary_food_first_day = df_dietary_food_first_day.rename(columns={
    'SEQN': 'seqn_no',
    'DR1CCMNM': 'food_no',
    'DR1CCMTX': 'food_type',
    'DR1FS': 'food_source',
    'DR1IGRMS': 'food_gms',
    'DR1IKCAL': 'energy_kcal'
})

# Step 3: Changing the data type of 'seqn_no' to int
df_dietary_food_first_day['seqn_no'] = df_dietary_food_first_day['seqn_no'].astype(int)
```

```python
# Group by 'seqn_no' and sum 'food_gms' and 'energy_kcal' for each participant
agg_df_01 = df_dietary_food_first_day.groupby('seqn_no')[['food_gms', 'energy_kcal']].sum().reset_index()
```

```python
agg_df_01.columns
```

Export as csv

```python
#export Day1 food as csv to data/dietary_data

file_path = "../data/dietary_data/dietary_food_data_day_01.csv" 

try:
    df_dietary_food_first_day.to_csv(file_path, index=False, encoding='utf-8')
    print(f"DataFrame successfully saved to: {file_path}")
except Exception as e:
    print(f"Error saving DataFrame to CSV: {e}")
```

```python
# Now building the URL with the values from the .env file
url = f'postgresql://{pg_user}:{pg_pass}@{pg_host}:{pg_port}/{pg_db}'

engine = create_engine(url, echo=False) 
```

```python
# Creating table and adding Day1 food data to the database
df_dietary_food_first_day.to_sql(name = 'dietary_food_day_01', con=engine, schema='capstone_group_3',if_exists='replace',index=False)
```

```python
# Group by 'seqn_no' and sum 'food_gms' and 'energy_kcal' for each participant
agg_df_02 = df_dietary_food_first_day.groupby('seqn_no')[['food_gms', 'energy_kcal']].sum().reset_index()
agg_df_02
```

Calculating avg for food_gms and energy_Kcal for each seqn_no for both days

```python
# Now let's avg the food data on each seqn_no

# Concatenate the two dataframes
df_food = pd.concat([agg_df_01, agg_df_02])

# List of columns that need to be aggregated
food_columns = [
    'food_gms', 'energy_kcal'
]

# Select only the relevant nutrient columns and group by 'seqn_no'
df_food_filtered = df_food[['seqn_no'] + food_columns]

# Group by 'seqn_no' and calculate the mean for all nutrient columns
df_food_avg = df_food_filtered.groupby('seqn_no').agg('mean').reset_index()
```
