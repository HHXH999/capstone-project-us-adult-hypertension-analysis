Calculating avg for food_gms and energy_Kcal for each seqn_no for both days

```python
# Now let's avg the food data on each seqn_no

# Concatenate the two dataframes
df_food = pd.concat([agg_df_01, agg_df_02])

# List of columns that need to be aggregated
food_columns = [
    'food_gms', 'energy_kcal'
]

# Select only the relevant nutrient columns and group by 'seqn_no'
df_food_filtered = df_food[['seqn_no'] + food_columns]

# Group by 'seqn_no' and calculate the mean for all nutrient columns
df_food_avg = df_food_filtered.groupby('seqn_no').agg('mean').reset_index()
```

# The recommended Energy in Kcalories for differnt ages and differnt gender is represented in a list

```python
calorie_recommendations_list = [
    {"Group": "Women (19-30)", "Sedentary": "1,800-2,000", "Moderately Active": "2,000-2,200", "Active": "2,400"},
    {"Group": "Women (31-50)", "Sedentary": "1,800", "Moderately Active": "2,000", "Active": "2,200"},
    {"Group": "Women (51+)",   "Sedentary": "1,600", "Moderately Active": "1,800", "Active": "2,000-2,200"},
    {"Group": "Men (19-30)",   "Sedentary": "2,400", "Moderately Active": "2,600-2,800", "Active": "3,000"},
    {"Group": "Men (31-50)",   "Sedentary": "2,200-2,400", "Moderately Active": "2,600", "Active": "2,800-3,000"},
    {"Group": "Men (51+)",     "Sedentary": "2,000-2,200", "Moderately Active": "2,400-2,600", "Active": "2,800"}
]

# Converting the list of dictionaries to a data frame 
df_calories = pd.DataFrame(calorie_recommendations_list)
```

```python
df_calories
```

```python
# ALTER DEFAULT PRIVILEGES IN SCHEMA capstone_group_3
# GRANT SELECT, INSERT, UPDATE, DELETE ON TABLES TO zhenyang;
```

```python
# Step 1: Melt the activity levels into rows
df_calories_melted = df_calories.melt(id_vars='Group', var_name='ActivityLevel', value_name='RecommendedRange')

# Step 2: Extract min and max values from range
def parse_range(value):
    # --- FIX 1: Remove commas before splitting/converting ---
    value = value.replace(',', '')
    if '-' in value:
        min_val, max_val = map(int, value.split('-'))
    else:
        min_val = max_val = int(value)
    return pd.Series([min_val, max_val])

df_calories_melted[['Min_kcal', 'Max_kcal']] = df_calories_melted['RecommendedRange'].apply(parse_range)


# Step 3: Merge actual data with recommendation table
df_compare = pd.merge(df_food_avg, df_calories_melted, how='left', on=['seqn_no'])

# Step 4: Compare actual kcal to recommended range
df_compare['Below Range'] = df_compare['energy_kcal'] < df_compare['Min_kcal']
df_compare['Above Range'] = df_compare['energy_kcal'] > df_compare['Max_kcal']
df_compare['Within Range'] = (~df_compare['Below Range']) & (~df_compare['Above Range'])

# Optional: final result
df_result = df_compare[['seqn_no', 'Group', 'ActivityLevel', 'energy_kcal', 'Min_kcal', 'Max_kcal', 'Within Range']]

print(df_result)
```
