## Dietary food data for Day 01

```python
# Exporting the Dietary food day 1 data
df_dietary_food_first_day = pd.read_sas('F:\Susanna Course\Capstone_data\DR1IFF_L.xpt', format="xport", encoding="utf-8")
```

```python
df_dietary_food_first_day.head(5)
```

```python
df_dietary_food_first_day.shape
```

```python
# Renaming the Columns
df_dietary_food_first_day = df_dietary_food_first_day.rename(columns={'SEQN': 'seqn_no', 'WTDRD1': 'survey_wt_day_01','WTDR2D': 'survey_wt_day_02','DR1ILINE':'food_line_no','DR1DRSTZ':'dietary_recall_status','DR1EXMER':'interview_code','DR1CCMNM':'food_no','DR1CCMTX':'food_type','DR1FS':'food_source','DR1IGRMS':'food_gms'})
```

```python
df_dietary_food_first_day
```

```python
# Keeping the columns which has meaningful values
df_dietary_food_first_day
columns_to_keep = ['seqn_no','survey_wt_day_01', 'survey_wt_day_02', 'food_line_no','dietary_recall_status','food_gms']

df_dietary_filtered_day_01 = df_dietary_food_first_day[columns_to_keep]
```

```python
 # The filtered day_o1 foods Dietary data
df_dietary_filtered_day_01
```

```python
# Group by 'seqn_no' and sum the 'food_gms' for each participant
agg_df = df_dietary_filtered_day_01.groupby('seqn_no')['food_gms'].sum().reset_index()
```

```python
agg_df
```

The average daily calorie needs vary widely:
Men: 2,000 to 2,500 calories per day.
Women: 1,800 to 2,200 calories per day.
Children and Adolescents: Their needs range from 1,200 to 2,500 calories depending on age and activity level.
Physically active individuals: May require 2,500–3,000 calories or more.
These calories are distributed across macronutrients like carbohydrates, proteins, and fats.

```python
#export as csv

file_path = "../data/dietary_data/dietary_data_day_01.csv" 

try:
    df_dietary_filtered_day_01.to_csv(file_path, index=False, encoding='utf-8')
    print(f"DataFrame successfully saved to: {file_path}")
except Exception as e:
    print(f"Error saving DataFrame to CSV: {e}")
```

```python
from dotenv import dotenv_values

config = dotenv_values()

# define variables for the login
pg_user = config['POSTGRES_USER']  # align the key label with your .env file !
pg_host = config['POSTGRES_HOST']
pg_port = config['POSTGRES_PORT']
pg_db = config['POSTGRES_DB']
pg_schema = config['POSTGRES_SCHEMA']
pg_pass = config['POSTGRES_PASS']

# Now building the URL with the values from the .env file
url = f'postgresql://{pg_user}:{pg_pass}@{pg_host}:{pg_port}/{pg_db}'

engine = create_engine(url, echo=False) 
```

```python
# creating table in the database 

df_dietary_filtered_day_01.to_sql(name = 'dietary_food_data_01', con=engine, schema='capstone_group_3',if_exists='replace',index=False)
```

## Dietary Nutrient Intake data for Day 01

```python
# Exporting the Dietary Nutrients intake day  data
df_dietary_nutrient_intake_day_01 = pd.read_sas('F:\Susanna Course\Capstone_data\DR1TOT_L.xpt', format="xport", encoding="utf-8")
```

```python
df_dietary_nutrient_intake_day_01.shape
```

```python
# List of relevant nutrient columns which could potentially be related to the research question
nutrient_columns = [
    'SEQN',
    'DR1TKCAL', 'DR1TPROT', 'DR1TCARB','DR1TSUGR','DR1TFIBE', 'DR1TTFAT',
     'DR1TSFAT','DR1TMFAT','DR1TPFAT','DR1TCHOL','DR1TSODI','DR1TPOTA','DR1TMOIS',
    'DR1TVC','DR1TVD','DR1TVK','DR1TCALC'
    ]
```

```python
# Filter the dataframe to keep only the relevant columns
df_nutrients_day01 = df_dietary_nutrient_intake_day_01[nutrient_columns]
```

```python
# Rename columns
df_nutrients  = df_nutrients.rename(
    columns={'SEQN': 'seqn_no',
             'DR1TKCAL':'energy_kcal','DR1TPROT':'protein_gms','DR1TCARB':'carbohydrate_gms','DR1TTFAT':'total_fat_gms',
             'DR1TSUGR': 'total_sugars_gms','DR1TFIBE':'dietary_fiber_gms','DR1TSFAT':'saturated_fats_gms', 'DR1TMFAT':'mono_saturated_fats_gms',
             'DR1TPFAT':'poly_saturated_fats_gms','DR1TCHOL':'cholestrol_gms','DR1TCALC':'calcium_mg',
             'DR1TSODI':'sodium_mg','DR1TPOTA':'potassium_mg','DR1TCAFF':'caffeine_mg','DR1TALCO':'alcohol_gms',
              'DR1TMOIS':'moisture_gms','DR1TVC':'vitaminC_mg','DR1TVD':'vitaminD_mcg','DR1TVK':'vitaK_mcg'})
```

```python
df_nutrients_day01.info()
```

```python
# Now changing the data type of 'seqn_no' to int
df_nutrients_day01['seqn_no'] = df_nutrients_day01['seqn_no'].astype(int)
```

```python
# Display the first few rows to check the data
df_nutrients_day01.head()
```

```python
#export as csv

file_path = "../data/dietary_data/dietary_nutrient_data_01.csv" 

try:
    df_nutrients_day01.to_csv(file_path, index=False, encoding='utf-8')
    print(f"DataFrame successfully saved to: {file_path}")
except Exception as e:
    print(f"Error saving DataFrame to CSV: {e}")
```

```python
df_nutrients_day01.to_sql(name = 'dietary_nutrient_data_01', con=engine, schema='capstone_group_3',if_exists='replace',index=False)
```
