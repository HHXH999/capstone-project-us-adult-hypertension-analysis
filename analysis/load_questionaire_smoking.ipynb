```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine, types # Export to DataBase

# read XPT file - Smoking - Cigarette Use (first-hand)
df_smoking = pd.read_sas("../data/questionaire_data/SMQ_L.xpt", format="xport", encoding="utf-8")
df_smoking
```

```python
print("Smoking Questionaire Data Info:")
df_smoking.info() 

# SMQ020 - Smoked at least 100 cigarettes in life
# SMQ040 - Do you now smoke cigarettes?

# SMD641 - # days smoked cigs during past 30 days
# SMD650 - Avg # cigarettes/day during past 30 days
# SMD100MN - Cigarette Menthol indicator
# SMQ621 - Cigarettes smoked in entire life
# SMD630 - Age first smoked whole cigarette
# SMAQUEX2 - Questionnaire Mode Flag -> drop 
```

```python
# Define the columns you want to keep and their new names
smoking_columns_to_keep_and_rename = {
    'SEQN': 'Participant_ID',                     # Keep SEQN for merging
    'SMQ020': 'Smoked_100_Cigs_Lifetime',          # New name for "Smoked at least 100 cigarettes in life"
    'SMQ040': 'Smokes_Now'                         # New name for "Do you now smoke cigarettes?"
}

# Select only the desired columns from the original DataFrame
df_smoking_selected = df_smoking[list(smoking_columns_to_keep_and_rename.keys())].copy()

# Rename the columns in the new DataFrame
df_smoking_selected.rename(columns=smoking_columns_to_keep_and_rename, inplace=True)


# --- Verification Steps ---
print("--- Selected and Renamed Smoking Data Info ---")
df_smoking_selected.info()

print("\n--- First few rows of the new DataFrame ---")
print(df_smoking_selected.head())
```

```python
import pandas as pd
import numpy as np

try:
    df_smoking_2nd = pd.read_sas("../data/questionaire_data/SMQFAM_L.xpt", format="xport", encoding="utf-8")
    print("--- Data loaded successfully. Initial data preview: ---")
    print(df_smoking_2nd.head())
except FileNotFoundError:
    print("Error: File not found. Please check your file path.")
    exit()
```

```python
# Step 1: Replace invalid small float values
df_smoking_2nd['SMD460'] = pd.to_numeric(df_smoking_2nd['SMD460'], errors='coerce')
df_smoking_2nd['SMD470'] = pd.to_numeric(df_smoking_2nd['SMD470'], errors='coerce')

# Step 2: Round if necessary and cast to Int64 to support NA
df_smoking_2nd['SMD460'] = df_smoking_2nd['SMD460'].round().astype('Int64')
df_smoking_2nd['SMD470'] = df_smoking_2nd['SMD470'].round().astype('Int64')

# Step 3:  Rename for clarity
df_smoking_2nd.rename(columns={
    'SEQN': 'Participant_ID',
    'SMD460': 'Household_Smokers_Total',
    'SMD470': 'Household_Smokers_Indoors'
}, inplace=True)

# Step 4: Inspect again
print("Second-Hand Smoking Questionaire Data Info:")
df_smoking_2nd.info() 
```

```python
df_smoking_2nd
```

```python
# --- Merge the two DataFrames ---
# Use pd.merge to combine them based on the common 'Participant_ID' column.

df_smoking_combined = pd.merge(
    df_smoking_selected,
    df_smoking_2nd,
    on='Participant_ID',  # The common column to merge on
    how='outer'           # The type of merge to perform
)

# --- Verification ---
print("--- Combined Smoking Data Info ---")
df_smoking_combined.info()

print("\n--- First few rows of Combined Smoking Data ---")
print(df_smoking_combined.head())
```

```python
# To see how the merge handled different participants, you can check specific rows
print(df_smoking_combined[df_smoking_combined['Participant_ID'] == 130380])
```

```python
df_smoking_combined = df_smoking_combined.convert_dtypes() 
```

```python
# export as csv
file_path = "../data/questionaire_data/cleaned_smoking_combined.csv" 

try:
    df_smoking_combined.to_csv(file_path, index=False, encoding='utf-8')
    print(f"DataFrame successfully saved to: {file_path}")
except Exception as e:
    print(f"Error saving DataFrame to CSV: {e}")
```

```python
from dotenv import dotenv_values

config = dotenv_values()

# define variables for the login
pg_user = config['POSTGRES_USER']  # align the key label with your .env file !
pg_host = config['POSTGRES_HOST']
pg_port = config['POSTGRES_PORT']
pg_db = config['POSTGRES_DB']
pg_schema = config['POSTGRES_SCHEMA']
pg_pass = config['POSTGRES_PASS']

# Now building the URL with the values from the .env file
url = f'postgresql://{pg_user}:{pg_pass}@{pg_host}:{pg_port}/{pg_db}'

engine = create_engine(url, echo=False) 
```

```python
df_smoking_combined.to_sql(name = 'smoking_combined_questionaire_data', con=engine, schema='capstone_group_3',if_exists='replace',index=False)
```
